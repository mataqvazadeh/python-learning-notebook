## Templates

### Folder Structure

Place templates inside a folder named after the app, inside a `templates/` directory:
```
myapp/
  templates/
    myapp/
      home.html
```

### Passing Context

Pass data to a template using the `context` dict:
```python
def home(request):
    return render(request, 'myapp/home.html', {'title': 'Welcome'})
```

In the template:
```html
<h1>{{ title }}</h1>
```

### Template Tags

**for loop:**
```html
{% for item in items %}
  <li>{{ item }}</li>
{% endfor %}
```

**if statement:**
```html
{% if user.is_authenticated %}
  <p>Hello, {{ user.username }}</p>
{% endif %}
```


## URL Patterns

### Static URL
```python
path('about/', views.about, name='about'),
```

### URL with Parameter
```python
path('<int:question_id>/', views.detail, name='detail'),
```

### Regex URL
```python
from django.urls import re_path
re_path(r'^articles/(?P<year>[0-9]{4})/$', views.year_archive),
```

## Defining Models: Basic Model and Field Types

```python
from django.db import models

class Article(models.Model):
    title = models.CharField(max_length=200)
    content = models.TextField()
    created_at = models.DateTimeField(auto_now_add=True)
```

Common field types:
- `CharField` — short text with a max length
- `TextField` — unlimited text
- `IntegerField` — whole numbers
- `DateTimeField` — date and time
- `BooleanField` — True/False


## Summary

- Templates use `{{ variable }}` for output and `{% tag %}` for logic.
- Template files live in `templates/<app_name>/` inside the app directory.
- URL patterns map URLs to view functions using `path()` or `re_path()`.
- `<int:pk>` and named regex groups capture URL segments as view arguments.
- Create a superuser with `python manage.py createsuperuser` to access the admin.
