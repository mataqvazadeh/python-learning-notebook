# Sessions & Cookies

## HTTP is Stateless

HTTP does not retain information between requests. Each request is independent. To track users across requests, Django provides cookies and sessions.


## Cookies

Cookies store small pieces of data on the client (browser).

### Setting a Cookie
```python
def set_theme(request):
    response = HttpResponse("Theme saved.")
    response.set_cookie('theme', 'dark', max_age=3600)
    return response
```

### Reading a Cookie
```python
def get_theme(request):
    theme = request.COOKIES.get('theme', 'light')
    return HttpResponse(f"Current theme: {theme}")
```

### Use Cookies For
- Non-sensitive preferences: theme, language, UI settings.


## Sessions

Sessions store data on the server. A session ID is sent to the client as a cookie, which the server uses to look up the stored data.

### Storing Data in a Session
```python
def add_to_cart(request, product_id):
    cart = request.session.get('cart', [])
    cart.append(product_id)
    request.session['cart'] = cart
    return HttpResponse("Added to cart.")
```

### Reading Session Data
```python
def view_cart(request):
    cart = request.session.get('cart', [])
    return HttpResponse(f"Cart: {cart}")
```

### Use Sessions For
- Sensitive data: authentication status, cart contents, user preferences that must not be exposed.


## Authentication vs Authorization

These two concepts are often confused:

| Concept | Question | Example |
|--------|----------|---------|
| Authentication | Are you who you claim to be? | Login with username + password |
| Authorization | What are you allowed to do? | Can this user delete a post? |

Django handles both: `django.contrib.auth` provides authentication, and the permission system handles authorization.


## Login Form Example

A basic HTML login form using Django's CSRF protection:

```html
<form method="post">
  {% csrf_token %}
  <input name="username" placeholder="Username"><br>
  <input name="password" type="password" placeholder="Password"><br>
  <button type="submit">Login</button>
</form>
{% if error %}
  <p style="color:red">{{ error }}</p>
{% endif %}
```

The `{% csrf_token %}` tag inserts a hidden token that Django uses to protect against Cross-Site Request Forgery attacks.


## Summary

- HTTP is stateless; cookies and sessions add state across requests.
- Cookies are stored on the client — use for non-sensitive data only.
- Sessions are stored on the server — use for sensitive data like authentication.
- Authentication verifies identity; authorization controls what a user can do.
- Always include `{% csrf_token %}` in POST forms to prevent CSRF attacks.
