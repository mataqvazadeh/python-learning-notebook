# Django Model Fields & Relationships

## Additional Field Types

### DateField and BooleanField
```python
class Event(models.Model):
    name = models.CharField(max_length=100)
    date = models.DateField()
    is_active = models.BooleanField(default=True)
```

### Choices
The `choices` argument restricts a field to a defined set of values:
```python
class Order(models.Model):
    STATUS_CHOICES = [
        ('pending', 'Pending'),
        ('shipped', 'Shipped'),
        ('delivered', 'Delivered'),
    ]
    status = models.CharField(max_length=20, choices=STATUS_CHOICES, default='pending')
```

Use `get_FOO_display()` to get the human-readable label:
```python
order = Order.objects.get(pk=1)
order.get_status_display()  # → 'Pending'
```


## Relationships

### OneToOneField
Each instance of one model is related to exactly one instance of another:
```python
class Profile(models.Model):
    user = models.OneToOneField(User, on_delete=models.CASCADE)
    bio = models.TextField(blank=True)
```

### ManyToManyField
An instance of one model can relate to many instances of another and vice versa:
```python
class Post(models.Model):
    tags = models.ManyToManyField('Tag', blank=True)
```

### related_name
`related_name` sets the reverse accessor on the related model:
```python
class Comment(models.Model):
    post = models.ForeignKey(Post, on_delete=models.CASCADE, related_name='comments')
```
Access from the post side: `post.comments.all()`


## The Meta Class

Use the inner `Meta` class to configure model-level options:

```python
class Article(models.Model):
    title = models.CharField(max_length=200)
    published_at = models.DateTimeField()

    class Meta:
        ordering = ['-published_at']       # newest first by default
        verbose_name = 'Article'
        verbose_name_plural = 'Articles'
        unique_together = [['title', 'published_at']]
```

- `ordering` — default sort order for querysets
- `verbose_name` / `verbose_name_plural` — human-readable names shown in admin
- `unique_together` — enforces that the combination of fields is unique


## Using Queries

```python
# filter — returns a QuerySet
Post.objects.filter(status='published')

# exclude — returns everything that does NOT match
Post.objects.exclude(status='draft')

# get — returns exactly one object (raises if 0 or >1)
Post.objects.get(pk=5)

# first — returns the first result or None
Post.objects.filter(status='published').first()
```


## Summary

- `DateField`, `BooleanField`, and `choices` cover common data patterns.
- `get_FOO_display()` returns the human-readable label for a choices field.
- Relationships: `OneToOneField`, `ForeignKey`, `ManyToManyField`.
- `related_name` controls the reverse accessor on the related model.
- The `Meta` class configures ordering, verbose names, and unique constraints.
- Core query methods: `filter()`, `exclude()`, `get()`, `first()`.
